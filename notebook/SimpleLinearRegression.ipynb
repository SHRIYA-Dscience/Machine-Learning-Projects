# SIMPLE LINEAR REGRESSION 

The simple linear regression is supervised machine learning algorithm which is used to find out relationship between single independent variable and depenedent varaible. 


**About Dataset**
The table below gives the heights of fathers and their sons, based on a famous experiment by Karl Pearson around 1903. The number of cases is 1078. Random noise was added to the original data, to produce heights to the nearest 0.1 inch.

Reference: https://www.kaggle.com/datasets/fundal/pearsons-height-data-simple-linear-regression


**Goal**

We take father height as a independ variable which is used to predict the son height as a dependent varaible.

**Columns**

1. Father - Height of father 
2. Son - Height of son
